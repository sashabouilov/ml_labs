````python
df = pd.read_csv('data.csv')
````

1. Какой средний вес людей представленных в наборе данных?

````python
average_weight = df['Weight (kg)'].mean()

print(average_weight)
````

2. Сбалансировано ли соотношение мужчин и женщин в наборе данных?

````python
gender_balanse = df['Gender'].value_counts(normalize=True) * 100

print(gender_balanse)
````

3. Правда ли, что люди с высоким уровнем стресса имеют более высокое кровяное давление, чем те, кто не подвержен стрессу

````python
df[['Systolic', 'Diastolic']] = df['Blood Pressure (s/d)'].str.split('/', expand=True).astype(float)
stress_to_pressure = df['Stress Levels'].corr(df['Systolic'])
print('Если корреляция положительная и близка к 1, значит, с ростом стресса давление действительно растет')
print(f'Корреляция: {stress_to_pressure}')
````

4. Правда ли, что уровень холестерина у мужчин и женщин отличается?

````python
hol_m = df.groupby('Gender')['Cholesterol Level (mg/dL)'].agg(['mean', 'median']).round(2)
print(hol_m)
````

5. Правда ли, что с возрастом женщины имеют свойство оценивать свои когнитивные способности хуже, чем в юности (порог юности - 30 лет, женщина в возрасте - 60 лет+)

````python
women = df[df['Gender'] == 'Female']
women_youth = women[women['Age (years)'] < 30]['Cognitive Function'].dropna()
women_senior = women[women['Age (years)'] >= 60]['Cognitive Function'].dropna()
print(f"Средняя самооценка (до 30 лет): {women_youth.mean():.2f}")
print(f"Средняя самооценка (60+ лет): {women_senior.mean():.2f}")
````

6. Влияет ли физическая активность на вес в возрасте более 50 лет?

````python
df_over_50 = df[df['Age (years)'] > 50]
weight_stats = df_over_50.groupby('Physical Activity Level')['Weight (kg)'].agg(['mean', 'median', 'count']).round(2)
print(weight_stats)
````

7. Сравните минимальный вес мужчин с высоким и низким заработком

````python
men = df[df['Gender'] == 'Male']
min_weight_high = men[men['Income Level'] == 'High']['Weight (kg)'].min()
min_weight_low = men[men['Income Level'] == 'Low']['Weight (kg)'].min()
print(f"Минимальный вес мужчин с высоким доходом: {min_weight_high} кг")
print(f"Минимальный вес мужчин с низким доходом: {min_weight_low} кг")
````

8. Правда ли, что у людей с индексом массы тела больше 75го перцентиля встречаются хронические заболевания чаще, чем у людей с индесом массы тела в переделах от 30 до 60 перцентиля?

````python
bmi_q30 = df['BMI'].quantile(0.30)
bmi_q60 = df['BMI'].quantile(0.60)
bmi_q75 = df['BMI'].quantile(0.75)
group_high_bmi = df[df['BMI'] > bmi_q75]
group_mid_bmi = df[(df['BMI'] >= bmi_q30) & (df['BMI'] <= bmi_q60)]
freq_high_str = (group_high_bmi['Chronic Diseases'] != 'None').mean() * 100
freq_mid_str = (group_mid_bmi['Chronic Diseases'] != 'None').mean() * 100
print(f"Доля с заболеваниями (высокий BMI больше 75): {freq_high_str:.2f}%")
print(f"Доля с заболеваниями (средний BMI от 30 до 60): {freq_mid_str:.2f}%")
````

9. Правда ли, что люди с высоким уровнем дохода имеют больший уровень физической активности, чем люди с низким доходом

````python
activity_by_income = pd.crosstab(df['Income Level'], df['Physical Activity Level'], normalize='index') * 100
print(activity_by_income.round(2))
````

10. Какая доля мужчин с высокой частотой пребывания на солнце имеет высокий заработок?

````python
men = df[df['Gender'] == 'Male']
sun_threshold = men['Sun Exposure'].quantile(0.75)
men_high_sun = men[men['Sun Exposure'] >= sun_threshold]
total_men_high_sun = len(men_high_sun)
high_income_men = len(men_high_sun[men_high_sun['Income Level'] == 'High'])
percentage = (high_income_men / total_men_high_sun) * 100
print(percentage)
````

### 3. Визуализация

Используя библиотеки matplotlib/sns/plotly/pandas и набор данных представленный ранее визуализируйте

1. Визуализируйте частоту встречаемых хронических заболеваний

````python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('data.csv')

diseases = df['Chronic Diseases'].dropna().astype(str)
diseases = diseases[diseases.str.lower() != 'none']

exploded_diseases = diseases.str.split(',').explode()
disease_counts = exploded_diseases.value_counts()

plt.figure(figsize=(10, 6))
sns.set_theme(style="whitegrid")
sns.barplot(x=disease_counts.values, y=disease_counts.index, palette='viridis')

plt.title('Частота встречаемости конкретных хронических заболеваний', fontsize=14)
plt.xlabel('Количество случаев', fontsize=12)
plt.ylabel('Заболевание', fontsize=12)

for i, v in enumerate(disease_counts.values):
    plt.text(v + 0.5, i, str(v), va='center', fontsize=11)

plt.show()
````

2. Как связана оценка собственных когнитивных функций с уровнем стресса?

````python
correlation = df['Cognitive Function'].corr(df['Stress Levels'])
print(f"Коэффициент корреляции: {correlation:.3f}")

plt.figure(figsize=(10, 6))
sns.set_theme(style="whitegrid")

sns.regplot(
    data=df, 
    x='Stress Levels', 
    y='Cognitive Function',
    scatter_kws={'alpha': 0.5, 'color': 'steelblue'},
    line_kws={'color': 'red'}
)

plt.title('Связь уровня стресса и самооценки когнитивных функций', fontsize=14)
plt.xlabel('Уровень стресса (0-100)', fontsize=12)
plt.ylabel('Когнитивные функции (0-100)', fontsize=12)
plt.xlim(0, 15)
plt.ylim(0, 100)
plt.show()
````

4. Постройте box plot отображающий распределение уровня стресса у людей с различным уровнем дохода

5. Постройте круговую диаграмму отображающую как ментальное здоровье связано с паттернами сна

````python
df['Combined'] = df['Sleep Patterns'] + ' + ' + df['Mental Health Status']
counts = df['Combined'].value_counts()

plt.figure(figsize=(12, 8))

plt.pie(
    counts, 
    labels=counts.index, 
    autopct='%1.1f%%', 
    startangle=140, 
    colors=plt.cm.Pastel1.colors,
    wedgeprops={'edgecolor': 'white', 'linewidth': 1}
)

plt.title('Связь паттернов сна и ментального здоровья', fontsize=16, pad=20)
plt.axis('equal')
plt.tight_layout()
plt.show()
````
